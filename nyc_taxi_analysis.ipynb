```python
class TaxiAnalysis:
    def __init__(self, data):
        self.data = data

    def average_speed(self):
        speed = self.data[:,7] / (self.data[:,8] / 3600)
        return speed.mean()

    def total_revenue(self):
        return self.data[:,-3].sum()

    def highest_fare(self):
        return self.data[:,10].max()
```

```python
analysis = TaxiAnalysis(taxi)

print(analysis.average_speed())
print(analysis.total_revenue())
print(analysis.highest_fare())
```
